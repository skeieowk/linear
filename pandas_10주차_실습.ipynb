### 2. 엑셀 파일로 저장하기

```python
import pandas as pd

names = ['철수', '영이', '길동', '미영', '순이', '철이']
lectures = ['국어', '영어', '수학']
scores = [[84,87,78],[21,15,84],[87,84,76],[100,87,99],[59,99,59],[46,77,56]]

d1 = pd.DataFrame(scores, index=names, columns=lectures)
d1
```

```python
# 참고: 최신 pandas(2.x)에서는 to_excel/read_excel의 encoding 옵션이 제거되어 생략합니다.
d1.to_excel("scores.xlsx")
```

### 2. 엑셀 파일 읽어오기

### 문제 1. pop_in_seoul.xlsx
3번째 행을 header로 하여 2, 4, 7, 10, 14번째 열을 읽어 DataFrame 만들기

```python
pop_seoul = pd.read_excel('pop_in_seoul.xlsx', usecols=[1,3,6,9,13], header=2)
pop_seoul.head()
```

column명을 ['구별', '인구수', '한국인', '외국인', '고령자']로 변경하기

```python
pop_seoul.columns = ['구별','인구수','한국인','외국인','고령자']
pop_seoul.head()
```

'구별'을 인덱스로 설정하기

```python
pop_seoul.set_index('구별', inplace=True)
pop_seoul.head()
```

1번째 행(합계 데이터) 삭제하기

```python
pop_seoul.drop('합계', inplace=True)
pop_seoul.head()
```

각 구별 한국인의 비율 구하기

```python
pop_seoul['한국인비율(%)'] = (pop_seoul.한국인 / pop_seoul.인구수) * 100
pop_seoul.head()
```

각 구별 외국인의 비율 구하기

```python
pop_seoul['외국인비율(%)'] = (pop_seoul.외국인 / pop_seoul.인구수) * 100
pop_seoul.head()
```

각 구별 고령자의 비율 구하기

```python
pop_seoul['고령자비율(%)'] = (pop_seoul.고령자 / pop_seoul.인구수) * 100
pop_seoul.head()
```

인구수가 가장 많은 구부터 보이도록 정렬하기

```python
pop_seoul.sort_values(by='인구수', ascending=False)
```

인구수가 적은 순서대로 5개 구 출력하기

```python
pop_seoul.sort_values(by='인구수', ascending=True).head()
```

외국인 수 내림차순 정렬하기

```python
pop_seoul.sort_values(by='외국인', ascending=False)
```

고령자 비율 오름차순 정렬하기

```python
pop_seoul.sort_values(by='고령자비율(%)', ascending=True)
```

### 문제 2. crime_in_seoul.csv

csv 파일을 읽어서 DataFrame 만들기 (인코딩은 'euc-kr')

```python
crime_seoul = pd.read_csv('crime_in_seoul.csv', encoding='euc-kr')
crime_seoul
```

살인이 10건 이상 발생한 곳 데이터 추출하기

```python
d = crime_seoul[crime_seoul.죄종 == '살인']
d1 = d[d.발생검거 == '발생']
d1[d1.건수 >= 10]
```

살인의 발생에 대한 합계 구하기

```python
d = crime_seoul[crime_seoul.죄종 == '살인']
d1 = d[d.발생검거 == '발생']
sum(d1.건수)
```

살인의 검거에 대한 합계 구하기

```python
d = crime_seoul[crime_seoul.죄종 == '살인']
d2 = d[d.발생검거 == '검거']
sum(d2.건수)
```
