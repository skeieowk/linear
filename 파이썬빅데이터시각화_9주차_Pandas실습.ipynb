# 파이썬 빅데이터시각화 9주차. Pandas 활용
DataFrame의 이해와 생성 · 조작 · 분석 실습코드

```python
import pandas as pd

names = ['철수', '영이', '길동', '미영', '순이', '철이']
lectures = ['국어', '수학', '영어']
scores = [[84,87,78],[21,15,84],[87,84,76],[100,87,99],[59,99,59],[46,77,56]]

def make_d2():
    """실습마다 처음 상태의 d2를 새로 만들어 주는 함수"""
    return pd.DataFrame(scores, index=names, columns=lectures)
```

# 1강. DataFrame의 이해와 생성

## 1-2. Series를 이용하여 DataFrame 생성

```python
names1 = ['철수', '영이', '길동', '미영', '순이', '철이']
score1 = [84, 21, 87, 100, 59, 46]
names2 = ['길동', '철수', '영이', '철이', '순이', '미영']
score2 = [99, 87, 87, 84, 77, 15]

s1 = pd.Series(score1, index=names1)
s2 = pd.Series(score2, index=names2)
```

```python
d = pd.DataFrame()
d['국어'] = s1
d['영어'] = s2
d
```

## 1-3. 데이터를 직접 넣어 DataFrame 생성
### index와 columns 지정 없을 때

```python
scores = [[84,87,78],[21,15,84],[87,84,76],[100,87,99],[59,99,59],[46,77,56]]
d1 = pd.DataFrame(scores)
d1
```

### index와 columns 지정 있을 때

```python
names = ['철수', '영이', '길동', '미영', '순이', '철이']
lectures = ['국어', '수학', '영어']
d2 = pd.DataFrame(scores, index=names, columns=lectures)
d2
```

### dictionary를 사용할 때

```python
scoresWithLectures = {'수학':[84,21,87,100,59,46], '국어':[87,15,84,87,99,77], '영어':[78,84,76,99,59,56]}
d3 = pd.DataFrame(scoresWithLectures, index=names)
d3
```

# 2강. DataFrame 조작

## 2-1. index 변경

### `DataFrame객체.index = 새로운 index 배열/list`

```python
d2 = make_d2()
d2.index = ['학생1','학생2','학생3','학생4','학생5','학생6']
d2
```

### `rename(index={...})` → 새로운 DataFrame 객체 반환 (원본 유지)

```python
d2.rename(index={'학생1':'철수', '학생2':'영이'})
```

```python
d2  # 원본은 그대로
```

### `rename(index={...}, inplace=True)` → 원본 객체 변경

```python
d2.rename(index={'학생1':'철수', '학생2':'영이'}, inplace=True)
d2
```

## 2-2. column 변경

### `DataFrame객체.columns = 새로운 columns 배열/list`

```python
d2 = make_d2()
d2.columns = ['과목1','과목2','과목3']
d2
```

### `rename(columns={...})` → 새로운 DataFrame 객체 반환 (원본 유지)

```python
d2.rename(columns={'과목1':'국어', '과목2':'수학'})
```

```python
d2  # 원본은 그대로
```

### `rename(columns={...}, inplace=True)` → 원본 객체 변경

```python
d2.rename(columns={'과목1':'국어', '과목2':'수학'}, inplace=True)
d2
```

## 2-3. 열 순서 변경

```python
d2 = make_d2()
column_customed = ['영어','국어','수학']
d2[column_customed]
```

```python
columns_sorted = sorted(d2.columns)
d2[columns_sorted]
```

## 2-4. 행 삭제
### `DataFrame객체.drop(index/list/배열, [axis=0])` → 새로운 객체 반환

```python
d2 = make_d2()
d2.drop('철수', axis=0)
```

```python
d2.drop(['영이','미영'], axis=0)
```

```python
d2  # 원본은 그대로
```

### `inplace=True` → 원본 객체 변경

```python
d2.drop('영이', axis=0, inplace=True)
d2
```

## 2-5. 열 삭제
### `DataFrame객체.drop(column명/list/배열, axis=1)` → 새로운 객체 반환

```python
d2 = make_d2()
d2.drop('국어', axis=1)
```

```python
d2.drop(['수학','영어'], axis=1)
```

```python
d2  # 원본은 그대로
```

### `inplace=True` → 원본 객체 변경

```python
d2.drop('영어', axis=1, inplace=True)
d2
```

## 2-6. 행 추가
### `DataFrame객체.loc[새index명] = 데이터`  (예시 - 합계 추가)

```python
d2 = make_d2()
d2.loc['합계'] = d2.sum(axis=0)
d2
```

## 2-7. 열 추가
### `DataFrame객체[새column명] = 데이터`  (예시 - 합계 추가)
> 앞에서 '합계' 행을 추가한 d2에 이어서 실행하므로, 맨 아래 합계 행의 값(1298)은 위 행들의 합이 아니라 '합계' 행 자체를 포함한 값이 됩니다. (수업자료와 동일)

```python
d2['합계'] = d2.sum(axis=1)
d2
```

## 2-8. 행의 값 수정
### `DataFrame객체.loc[index명] = 데이터`

```python
d2 = make_d2()
d2.loc['철수'] = [80,90,80]
d2
```

### `DataFrame객체.iloc[index번호] = 데이터`

```python
d2.iloc[0] = [84,87,78]
d2
```

## 2-9. 열의 값 수정
### `DataFrame객체[column명] = 데이터`

```python
d2 = make_d2()
d2['국어'] = [80,30,90,100,60,50]
d2
```
