## 3강. 누락값, 중복 데이터 처리 및 시계열 데이터 활용

### 1. 누락 데이터 확인 (isnull / notnull)

```python
names1 = ['철수', '영이', '길동', '미영', '순이', '철이']
names2 = ['철수', '영이', '길동', '미영', '순이', '슬기']
lectures = ['국어', '영어', '수학']
scores1 = [[84,87,78],[21,15,84],[87,84,76],[100,87,99],[59,99,59],[46,77,56]]
scores2 = [[80,80,90],[50,60,70],[80,90,80],[100,90,99],[77,95,70],[55,80,65]]

d1 = pd.DataFrame(scores1, index=names1, columns=lectures)
d2 = pd.DataFrame(scores2, index=names2, columns=lectures)
display(d1)
display(d2)
```

```python
# 인덱스가 서로 다른 DataFrame끼리 더하면 한쪽에만 있는 행은 NaN이 된다
d3 = d1 + d2
d3
```

```python
# 누락 데이터이면 True
d3.isnull()
```

```python
# 유효한 데이터이면 True
d3.notnull()
```

### 2. 누락 데이터 제거 (dropna)

```python
# 행 제거: '국어'가 누락된 행 제거
d4 = d3.dropna(subset=['국어'])
d4
```

```python
# 참고) 열 제거 / thresh 옵션
# d3.dropna(axis=1, thresh=개수)    : 유효한 값이 thresh개 미만인 열 삭제
# d3.dropna(axis=0, thresh=개수)    : 유효한 값이 thresh개 미만인 행 삭제
# d3.dropna(how='all') / how='any' : 전부 / 하나라도 누락된 행 삭제
d3.dropna(thresh=1)
```

### 3. 누락 데이터 치환 (fillna)

```python
# '국어' 열의 누락값을 '국어' 최소값으로 치환
# (슬라이드의 fillna(..., inplace=True)는 최신 pandas에서 경고/미동작할 수 있어 대입 방식으로 작성)
min_kor = d3['국어'].min()
d3['국어'] = d3['국어'].fillna(min_kor)
d3
```

### 4. 중복 데이터 확인 (duplicated)

```python
# 행 중복 확인
d3.duplicated()
```

```python
# 열 중복 확인
d3['국어'].duplicated()
```

### 5. 중복 데이터 제거 (drop_duplicates)

```python
d4 = d3.drop_duplicates()
d4
```

```python
d5 = d3.drop_duplicates(subset=['국어','영어'])
d5
```

### 6. 다른 자료형을 시계열 객체로 변환

문자열 → Timestamp (to_datetime)

```python
names = ['철수','영이','길동','미영']
df = pd.DataFrame({'생년월일': ['1990-03-02','1991-06-08','1990-11-22','1991-01-05'],
                   '점수': [90,95,80,88]}, index=names)
df
```

```python
df.info()
```

```python
df['생년월일'] = pd.to_datetime(df['생년월일'])
df
```

```python
df.info()
```

Timestamp → Period (to_period)

```python
dates = ['2000-01-01', '2001-03-02', '2002-12-31']
ts_dates = pd.to_datetime(dates)
ts_dates
```

```python
pr_day = ts_dates.to_period(freq='D')
pr_day
```

```python
pr_month = ts_dates.to_period(freq='M')
pr_month
```

```python
# 연 단위: 슬라이드의 'A'는 최신 pandas에서 'Y'로 바뀜
pr_year = ts_dates.to_period(freq='Y')
pr_year
```

### 7. 시계열 데이터 활용

날짜 데이터 분리 – dt.year, dt.month, dt.day

```python
df['년'] = df['생년월일'].dt.year
df['월'] = df['생년월일'].dt.month
df['일'] = df['생년월일'].dt.day
df
```

날짜 데이터 분리 – dt.to_period()

```python
df['년월'] = df['생년월일'].dt.to_period(freq='M')
df
```

날짜 인덱스 활용

```python
df2 = df.set_index('생년월일')
df2
```

```python
# 연도로 검색
df2.loc['1990']
```

```python
# 연-월로 검색
df2.loc['1990-03']
```

```python
# 기간으로 검색 (슬라이싱은 인덱스가 정렬되어 있어야 안정적이라 sort_index() 후 사용)
df2.sort_index().loc['1990-03-02':'1990-12-31']
```

```python
# 기준 날짜와의 차이
day = pd.to_datetime('1990-03-02')
df2['날짜_차이'] = df2.index - day
df2
```

### 8. 시계열 데이터 만들기

Timestamp 배열 – date_range()

```python
ts_ms = pd.date_range(start='2000-01-01',
                      end=None,
                      periods=6,
                      freq='MS',
                      tz='Asia/Seoul')
ts_ms
```

```python
# 2개월 간격(월말): 최신 pandas는 '2ME', 구버전은 '2M'
try:
    ts_m = pd.date_range(start='2000-01-01', end=None, periods=6, freq='2ME', tz='Asia/Seoul')
except ValueError:
    ts_m = pd.date_range(start='2000-01-01', end=None, periods=6, freq='2M', tz='Asia/Seoul')
ts_m
```

Period 배열 – period_range()

```python
pr_m = pd.period_range(start='2000-01-01',
                       end=None,
                       periods=3,
                       freq='2M')
pr_m
```

```python
# 시간 단위: 슬라이드의 'H'는 최신 pandas에서 'h'로 바뀜
pr_h = pd.period_range(start='2000-01-01',
                       end=None,
                       periods=3,
                       freq='h')
pr_h
```
